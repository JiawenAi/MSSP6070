```python
import numpy as np
import pandas as pd

#change the working directory to where I uploaded my data：
%pushd "/content/drive/MyDrive/MSSP607/WeeklyModules/Week04"

df = pd.read_csv("parsed.csv")
df.head()
df['parsed_place'].head() #the locations
df['mag'].head() #magnitude
df['magType'].head() #type of magnitude
df['tsunami'].head() #have tsunami or not (true 1 /false 0)
```

```python
#2a Find the 95th percentile of earthquake magnitude in Japan using magType of 'mb'
#find Japan earthquake，and magType is mb
japan_mb= df[df['parsed_place'].str.endswith('Japan') &
    (df['magType'] == 'mb')]

#find the 95th percentile
percentile_95 = np.percentile(japan_mb['mag'], 95)
print("The 95th percentile of earthquake magnitutde in japan is:", percentile_95)
```

```python
#2b Find the percentage of earthquakes in Indonesia that were coupled with tsunamis.

#find indonesia data
indonesia = df[df['parsed_place'].str.endswith('Indonesia')]

#find indonesia tsunami data
indonesia_tsunami = indonesia[indonesia['tsunami'] == 1]

percentage_indonesia_tsunami = (len(indonesia_tsunami) / len(indonesia)) * 100

print("The percentage of earthquakes in Indonesia that were coupled with tsunamis is:",
      f'{percentage_indonesia_tsunami:.2f}%')
```

```python
#2c Get summary statistics for earthquakes in Nevada
#find Nevada earthquakes data
nevada_data = df[df['parsed_place'].str.endswith('Nevada')]
nevada_data.describe()
```

```python
#2d Add a column to the dataframe
#indicating whether or not the earthquake happened in a country or US state that is on the Ring of Fire.
df['ring_of_fire'] = ''
df.columns

```

```python
df['ring_of_fire'] = df.parsed_place.str.contains(r'|'.join([
        'Bolivia', 'Chile', 'Ecuador', 'Peru', 'Costa Rica',
        'Guatemala', '^Mexico', 'Japan', 'Philippines', 'Indonesia',
        'New Zealand', 'Antarctic', 'Canada', 'Fiji',
        'Alaska', 'Washington', 'California', 'Russia',
        'Taiwan', 'Tonga', 'Kermadec Islands'
    ]))

df[['ring_of_fire', 'parsed_place']]
```

```python
#2e Calculate the number of earthquakes in the Ring of Fire locations and the number outside of them

df['ring_of_fire'].value_counts()
```

```python
ring_counts = df['ring_of_fire'].value_counts()
print('Number of earthquakes in the Ring of Fire:', ring_counts[True])
print('Number of earthquakes outside the Ring of Fire:', ring_counts[False])
```

```python
#2f Find the tsunami count along the Ring of Fire
fire_with_tsunami = df[(df['ring_of_fire'] == True) & (df['tsunami'] == 1)]

print('Number of Tsunami with the Ring of Fire:',len(fire_with_tsunami))
```
