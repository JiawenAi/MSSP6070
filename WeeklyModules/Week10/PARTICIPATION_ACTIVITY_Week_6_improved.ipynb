```python
# Load and preview the student performance dataset
df = pd.read_csv('data/Student Performance Data.csv')
df.head()
```

```python
# Check for missing values before transforming the data
df.isna().sum()
```

```python
# Create a new dataframe with numerical columns only and add the
# average score across math, reading, and writing using a lambda function
score_average = lambda x: (
    x.assign(
        average_score=x[['math score', 'reading score', 'writing score']]
        .mean(axis=1)
        .round(2)
    )
    .select_dtypes(include='number')
)

new_df = score_average(df)
new_df.head()
```
