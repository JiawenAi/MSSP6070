```python
#SP2 Faculty Profile Image Downloader
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://sp2.upenn.edu/sp2-directory/'

response = requests.get(url)
response.raise_for_status()

soup = BeautifulSoup(response.text, 'html.parser')
```

```python
#checking number of founds
links = soup.find_all('a', href=True)

print("Number of links:", len(links))
```

```python
#checking links found
images = soup.find_all('img')

print("Number of images found:", len(images))

for img in images[:10]:
    print(img.get('src'))
```

The first link appears to be the logo of SP2 website page. I think we should remove that from our profile picture downloading.

```python
#skip the first image (SP2 logo)
image_urls = [
    img.get('src')
    for img in images
    if img.get('src') is not None
]

profile_images = image_urls[1:6]
profile_images
```

```python
#the logo picture is removed
#try to download the first 5 pictures

#folder for downloaded
os.makedirs('faculty_images', exist_ok=True)

#download first 5 profile images
for i, image_url in enumerate(profile_images):

    image_response = requests.get(image_url)
    image_response.raise_for_status()

    file_name = f'faculty_images/faculty_{i+1}.jpg'

    with open(file_name, 'wb') as file:
        file.write(image_response.content)

    print('Downloaded:', file_name)
```
