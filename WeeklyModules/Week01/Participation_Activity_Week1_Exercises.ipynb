## Exercise 4: Generate the data by running this cell
This will give you a list of numbers to work with in the remaining exercises.

```python
import random
random.seed(0)
salaries = [round(random.random()*1000000, -3) for _ in range(100)]
print(list(salaries))
```

## Exercise 5: Calculating statistics and verifying

Use the data generated above to calulate in code the following statistics without importing the statistics module. Then use the statistics module to verify your results.  Import the statistics module from https://docs.python.org/3/library/statistics.html

### mean

```python
salaries_mean = sum(salaries)/len(salaries)
salaries_mean
print("Calculated mean:", salaries_mean)

import statistics
print("Statistics module mean:", statistics.mean(salaries))
```

### median

```python
sorted_salaries = sorted(salaries)
print(sorted_salaries[49])
print(sorted_salaries[50])
median_of_salaries = (sorted_salaries[49] + sorted_salaries[50])/2
print("Calculated meidan:", median_of_salaries)

print("Statistics module median:", statistics.median(salaries))
```

### mode

```python
salaries_mode = None
highest_count = 0

for salary in salaries:
    count = salaries.count(salary)

    if count > highest_count:
        highest_count = count
        salaries_mode = salary

print("Calculated mode:",salaries_mode)
print("Statistics module mode:", statistics.mode(salaries))
```

### sample variance
Remember to use Bessel's correction.

```python
#(xi​−xˉ)^2
squared_deviations = [(salary - salaries_mean) ** 2 for salary in salaries]
#∑(xi​−xˉ)^2
sum(squared_deviations)
#∑(xi​−xˉ)^2/(n−1)
sample_variance = sum(squared_deviations)/(len(salaries)-1)

print("Calculated sample variance:",sample_variance)

print("Statistics module sample variance:",statistics.variance(salaries))
```

### sample standard deviation
Remember to use Bessel's correction.

```python
sample_sd = sample_variance ** 0.5
print("Calculated sample standard deviation:",sample_sd)

print("Statistics module sample standard deviation:",statistics.stdev(salaries))
```

## Exercise 6: Calculating more statistics
### range

```python
salary_range = max(salaries)- min(salaries)
print("Calculated range:",salary_range)
```

### coefficient of variation

```python
#CV = sd/mean
cv = sample_sd/statistics.mean(salaries)
print("Coefficient of variation:",statistics.mean(salaries))
```

### interquartile range

```python
#Q1: median of the first half
first_half = sorted_salaries[:50]
Q1 = statistics.median(first_half)

#Q3:median of the later half
later_half = sorted_salaries[50:]
Q3 = statistics.median(later_half)

IQR = Q3 - Q1

print("Interquartile range:", IQR)

q1 = (sorted_salaries[24] + sorted_salaries[25]) / 2
q3 = (sorted_salaries[74] + sorted_salaries[75]) / 2
iqr = q3-q1
print("Checking IQR:",iqr)
```

### quartile coefficent of dispersion

```python
#𝑄𝐶𝐷=（𝑄3−𝑄1）/（𝑄3+𝑄1）
QCD = (Q3-Q1)/(Q3+Q1)
print ("Quartile coefficent of dispersion:", QCD)

```

## Exercise 7: Scaling data
### min-max scaling

```python
#𝑥𝑠𝑐𝑎𝑙𝑒𝑑=(𝑥−𝑚𝑖𝑛(𝑋))/𝑟𝑎𝑛𝑔𝑒(𝑋)
scaled_salaries = [
    (salary - min(salaries)) / salary_range
    for salary in salaries
]
print ("scaled salaries:",scaled_salaries)
```

### standardizing

```python
#𝑧𝑖=（𝑥𝑖−𝑥¯)/𝑠
z_values = [(salary - statistics.mean(salaries))/statistics.stdev(salaries)
for salary in salaries]

print ("Standardizing salaries:", z_values)
```

## Exercise 8: Calculating covariance and correlation
### covariance

```python
import numpy as np
cov_salaries = np.cov(scaled_salaries, z_values)
print('Covariance of scaled and standardized values of salaries:',cov_salaries)
```

### Pearson correlation coefficient ($\rho$)

```python
correlation_coefficient = cov_salaries/statistics.stdev(scaled_salaries)*statistics.stdev(z_values)
print('Correlation coefficient:',correlation_coefficient)
```
