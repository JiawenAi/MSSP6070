```python
#2b Read each file into a dataframe
aapl = pd.read_csv("data/Week 05 Examples/aapl.csv")
amzn = pd.read_csv("data/Week 05 Examples/amzn.csv")
fb = pd.read_csv("data/Week 05 Examples/fb.csv")
goog = pd.read_csv("data/Week 05 Examples/goog.csv")
nflx = pd.read_csv("data/Week 05 Examples/nflx.csv")
```

```python
#2c Add a "ticker" column to each dataframe
aapl["ticker"] = "AAPL"
amzn["ticker"] = "AMZN"
fb["ticker"] = "FB"
goog["ticker"] = "GOOG"
nflx["ticker"] = "NFLX"
```

```python
#2d Append each file together into a single dataframe
faang = pd.concat([aapl, amzn, fb, goog, nflx], ignore_index=True)


#2e save to a csv
faang.to_csv("data/Week 05 Examples/faang.csv", index=False)
```

```python
#2f change the 'date' column into a 'datetime'
faang = pd.read_csv("data/Week 05 Examples/faang.csv")
faang.dtypes
faang["date"] = pd.to_datetime(faang["date"])
# turn 'volume' column into integers
faang["volume"] = faang["volume"].astype(int)

#sort by 'date' and 'ticker.'
#Firstly, sort it with dates; if dates are the same, then sort it with ticker
faang = faang.sort_values(["date", "ticker"])
```

```python
#2g Find the seven rows within the highest value for volume.
faang.nlargest(7, "volume")
```

```python
#2h Use the melt() method to make the 'date' and 'ticker' key identifiers.
# Melt the remaining columns so that there are not separated.
melt_faang = faang.melt(
    id_vars=["date", "ticker"]
)
melt_faang.head(10)
```
