```python
#5
#Write a function that accepts the list 'spam' as an argument and returns a string with all the items separated by a comma and a space, with 'and' inserted before the last item. The function should work with any list.

spam = ['apples', 'bananas', 'tofu', 'cats']

def list_to_string(spam):
  result_string = ''
  for item in spam[:-1]: # dealing with all items except the last one
    result_string = result_string+ item + ', '

  result_string = result_string + 'and ' + spam[-1] #dealing with the last item
  return result_string

print(list_to_string(spam))

#test other list
print(list_to_string(letters))
```
